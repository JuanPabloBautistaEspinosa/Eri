## Pregunta de negocio (CRISP-DM)

**Problema:** Superstore vende productos de tres categorías, pero algunas ventas generan pérdidas fuertes (el Profit llega hasta -3,839.99). No sabemos qué las causa.

**Pregunta:** ¿Qué factores (descuento, categoría, subcategoría, región) hacen que una venta genere pérdida?

Fases de CRISP-DM:
1. **Entender el negocio:** reducir las pérdidas sin sacrificar ventas.
2. **Entender los datos:** 2,435 filas y 21 columnas, 4 nulos y 0 duplicados. El Profit promedio es 22.89.
3. **Preparar los datos:** tratar los nulos y convertir las fechas a formato fecha.
4. **Modelado:** usar un modelo para ver qué variables influyen más en el Profit.
5. **Evaluación:** revisar que los resultados respondan la pregunta y tengan sentido.
6. **Despliegue:** presentar conclusiones con gráficas o un reporte.

**Tipo de análisis:** primero **visualización** (por ejemplo, Profit vs. Discount) y después **aprendizaje supervisado**, porque hay una variable objetivo clara: el Profit.

## Tipo de estructura de los datos

1. El archivo CSV es **estructurado**, porque sus datos están organizados en filas y columnas.
2. Todas las filas tienen los mismos 21 campos, y cada columna tiene un tipo de dato definido (texto, número o fecha).
3. Si los datos vinieran en **JSON**, serían **semiestructurados**: tienen etiquetas, pero pueden tener niveles anidados y campos que cambian entre registros.
4. Habría que "aplanarlos" (por ejemplo, con `pd.json_normalize`) y revisar campos faltantes antes de analizarlos.
5. JSON es más flexible, pero daría más trabajo de preparación que un CSV.

## Tipo de procesamiento: Batch

Elegí **batch**, que consiste en procesar todos los datos juntos en un solo momento.

- Los datos de Superstore son **históricos** (ventas de 2015 y 2016); no llegan datos nuevos cada segundo.
- La pregunta de negocio es **estratégica** (qué provoca las pérdidas), así que no se necesitan respuestas inmediatas.
- El volumen es pequeño (2,435 filas) y se procesa en segundos.
- **Streaming** o **micro-batch** solo tendrían sentido para vigilar las ventas en vivo, por ejemplo para detectar descuentos anómalos en caja. Este caso no lo requiere.
