```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv('Sample - Superstore.csv', encoding='latin-1')
df.head()
```

```python
scaler = MinMaxScaler()
df [['Sales_normalized', 'Profit_normalized']] = scaler.fit_transform(df[['Sales', 'Profit']])
print(df[['Sales_normalized', 'Profit_normalized']])
```

```python
df['Profit_Margin'] = df['Profit'] / df['Sales']
print(df['Profit_Margin'])
```

```python
df ['Profilable'] = (df['Profit'] > 0).astype(int)
```

```python
ventas_cat = df.groupby('Category')['Sales'].agg(['sum', 'mean', 'count'])
print(ventas_cat)
```
